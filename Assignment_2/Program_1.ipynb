```python
import pandas as pd
import numpy as np

# Car dataset
data = {
    "Name": ["Mahindra Thar", "Maruti Swift", "Mahindra Scorpio", "Tata Punch", "Hyundai Creta"],
    "Min Price (Lakh)": [11.35, 6.49, 13.62, 6.13, 11],
    "Max Price (Lakh)": [17.6, 9.6, 17.42, 10.2, 20.15],
    "Range (kmpl)": [15.2, 24.8, 22, 20.09, 17.4],
    "CC": [1497, 1197, 2184, 1199, 1497],
    "Seats": [4, 5, 7, 5, 5],
    "Variants": [19, 11, 4, 25, 28],
    "Type": ["Petrol", "Petrol", "Diesel", "Petrol", "Petrol"],
}

df = pd.DataFrame(data)
df
```

```python
print("Type:", type(df))
print("Columns:", list(df.columns))
print("Index:", df.index.tolist())

df
```

```python
# Save to CSV, then load it back
df.to_csv("Car data.csv", index=False)
df_from_csv = pd.read_csv("Car data.csv")

# Save to Excel, then load it back
df.to_excel("Car data.xlsx", index=False)
df_from_excel = pd.read_excel("Car data.xlsx")

df_from_csv.tail()
```

```python
print("First 3 rows:")
display(df.head(3))

print("\nLast 3 rows:")
display(df.tail(3))

print("\nInfo:")
df.info()

print("\nDescribe:")
display(df.describe())

print("\nShape (rows, columns):", df.shape)
```

```python
# loc: label-based selection
print("loc example (rows 0-2, specific columns):")
display(df.loc[0:2, ["Name", "Type", "CC"]])

# iloc: position-based selection
print("\niloc example (first 3 rows, first 3 columns):")
display(df.iloc[0:3, 0:3])

# Boolean indexing: filter rows where Type is 'Petrol'
print("\nBoolean indexing (Type == 'Petrol'):")
display(df[df["Type"] == "Petrol"])
```
